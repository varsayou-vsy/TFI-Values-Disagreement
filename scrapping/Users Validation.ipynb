### All Together

```python
def ObtenerArchivos(ruta_actual,carpeta):
    """Esta función nos devuelve una lista con los archivos de una carpeta"""
    ruta_completa = os.path.join(ruta_actual,carpeta)
    archivos = glob(ruta_completa+"/*")
    return archivos
```

```python
files = ObtenerArchivos(r"//home/varsayou/Desktop/Clases/Proyecto/Value-disagreement/Datos/DEBAGREEMENT_data/",
                        carpeta="author_summary")
files
```

```python
lista = [] ### LISTA AUXILIAR --> 

for archivo in files:
    if ".csv" in archivo:
        print("archivo: ",archivo)
        df = pd.read_csv(archivo,
                         sep="|")
        lista.append(df)
        
df_all = pd.concat(lista,axis=0, sort = False)
df_all
```
